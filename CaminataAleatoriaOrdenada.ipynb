```r
set.seed(123)  # Para reproducibilidad
```

```r
# Generar números aleatorios
numerosAleatorios <- sample(1:99, 100, replace = TRUE)
```

```r
# Inicializar variables
resultados <- list()
totalExitos <- 0

cat("🔁 Iniciando simulación de caminatas aleatorias...\n\n")
```

```r
# Ejecutar 10 pruebas
for (numeroPrueba in 1:10) {
  posicionX <- 0
  posicionY <- 0
  movimientos <- c()

  cat(paste("📊 Prueba", numeroPrueba, "\n"))
  cat("Paso\tNúmero\tPosición (x, y)\n")

  for (paso in 1:10) {
    numero <- numerosAleatorios[(numeroPrueba - 1) * 10 + paso]

    if (numero < 25) {
      posicionX <- posicionX + 1
    } else if (numero < 50) {
      posicionX <- posicionX - 1
    } else if (numero < 75) {
      posicionY <- posicionY + 1
    } else {
      posicionY <- posicionY - 1
    }

    movimientos <- rbind(movimientos, c(paso, numero, posicionX, posicionY))
    cat(paste(paso, "\t", numero, "\t\t(", posicionX, ",", posicionY, ")\n"))
  }

  distanciaTaxi <- abs(posicionX) + abs(posicionY)
  resultadoExito <- ifelse(distanciaTaxi == 2, "ÉXITO", "NO")

  cat("✅ Resultado:", resultadoExito, "\n\n")
  resultados[[numeroPrueba]] <- list(datos = movimientos, exito = resultadoExito)

  if (resultadoExito == "ÉXITO") {
    totalExitos <- totalExitos + 1
  }
}
```

```r
# Mostrar resumen final
cat("📌 Resumen Final:\n")
cat(paste("Total de pruebas:", length(resultados), "\n"))
cat(paste("Éxitos:", totalExitos, "\n"))
cat(paste("Fracasos:", 10 - totalExitos, "\n"))
cat(paste("🎯 Probabilidad de éxito:", totalExitos / 10, "\n"))
```
